## Data Quality Assessment

The dataset is evaluated for common data-quality issues including:

- Missing values
- Duplicate records
- Duplicate customer identifiers
- Invalid categorical values
- Invalid binary values
- Unexpected numerical ranges

**Missing Values**

```python
missing_values = data.isnull().sum()
missing_values[missing_values > 0]
```

**Finding**

No missing values were identified in the dataset. Therefore, no
imputation or row removal is required.

**Duplicate Records**

```python
print("Duplicate rows:", data.duplicated().sum())
print("Duplicate Customer IDs:", data["CustomerId"].duplicated().sum())
```

**Finding**

No duplicate rows or duplicate Customer IDs were identified.The CustomerId check confirms that each customer is represented by a unique identifier.

**Data Type Validation**

```python
dtype_summary = pd.DataFrame({
    "Column": data.columns,
    "Data Type": data.dtypes.values
})

dtype_summary
```

**Finding**

Column data types are reviewed to ensure that numerical, categorical,
and binary variables are stored appropriately.

**Categorical & Binary Validation**

```python
data["Geography"].value_counts()
```

```python
data["Gender"].value_counts()
```

```python
binary_columns = [
    "HasCrCard",
    "IsActiveMember",
    "Exited"
]

for col in binary_columns:
    print(f"{col}: {sorted(data[col].unique())}")
```

**Finding**

- `Geography` contains three expected categories: France, Spain, and Germany.
- `Gender` contains two categories: Female and Male.
- `HasCrCard`, `IsActiveMember`, and `Exited` contain only valid binary values (0 and 1).

**Numerical Data Validation**

```python
numeric_columns = [
    "CreditScore",
    "Age",
    "Tenure",
    "Balance",
    "NumOfProducts",
    "EstimatedSalary"
]

data[numeric_columns].describe()
```

```python
print("Credit Score:", data["CreditScore"].min(), "-", data["CreditScore"].max())
print("Age:", data["Age"].min(), "-", data["Age"].max())
print("Tenure:", data["Tenure"].min(), "-", data["Tenure"].max())
print("NumOfProducts:", data["NumOfProducts"].min(), "-", data["NumOfProducts"].max())
```

**Finding**

Numerical variables are examined for their observed ranges and potential
business-logical anomalies.

## Business Anomaly Investigation

**Very low estimated salaries**

```python
low_salary = data[data["EstimatedSalary"] < 1000]

print(f"Customers with EstimatedSalary < 1000: {len(low_salary)}")

low_salary[
    ["CustomerId", "EstimatedSalary", "Age", "Balance", "Exited"]
]
```

**Finding**

A small number of customers have estimated salaries below 1,000.
These values are unusual but are not confirmed to be data-entry errors.
Therefore, they are retained rather than arbitrarily removed.

**Zero Balance**

```python
zero_balance = (data["Balance"] == 0).sum()
print(f"Customers with zero balance: {zero_balance:,}")
```

**Finding**

Zero account balances are considered valid observations and are therefore
retained in the dataset.

**Remove Irrelevant Columns**

```python
data = data.drop(columns=["RowNumber", "Surname"])
print(f"Dataset shape after column removal: {data.shape}")
```

**Finding**

`RowNumber` and `Surname` are removed because they do not provide useful
analytical information for customer churn analysis.

- `RowNumber` is a dataset-specific row identifier.
- `Surname` is a customer name and is not required for the analysis.

**Final Data Quality Validation**

```python
print("Dataset shape:", data.shape)
print("Duplicate rows:", data.duplicated().sum())
print("Duplicate Customer IDs:", data["CustomerId"].duplicated().sum())
print("Total missing values:", data.isnull().sum().sum())
```

```python
data.info()
```

**Finding**

The cleaned dataset is revalidated to confirm that the cleaning process
did not introduce missing values, duplicates, or unexpected data types.

**Export Clean Dataset**

```python
data.to_csv(
    "../data/cleaned/bank_customer_churn_cleaned.csv",
    index=False
)
```
